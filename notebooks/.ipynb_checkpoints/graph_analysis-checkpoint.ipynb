# Graph Analysis Notebook
This notebook demonstrates how to run Neo4j queries, convert results to DataFrames, and visualize graph data.

```python
import pandas as pd
from src.connect import get_driver
```

```python
driver = get_driver()
with driver.session() as session:
    result = session.run("MATCH (n) RETURN n LIMIT 10")
    records = [record.data() for record in result]
df = pd.DataFrame(records)
df.head()
```
